```python
# ===========================================
# PLANT DISEASE DETECTION - TASK 3 (FIXED)
# Image-Based Plant Leaf Disease Classification
# Rhombix Technologies Data Science Internship
# ===========================================

# ---------- 1. IMPORT LIBRARIES ----------
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
import cv2
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, models
from tensorflow.keras.preprocessing.image import ImageDataGenerator, load_img, img_to_array
from tensorflow.keras.applications import MobileNet
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint
from sklearn.metrics import classification_report, confusion_matrix
import warnings
warnings.filterwarnings('ignore')

print("="*60)
print("PLANT DISEASE DETECTION SYSTEM (FIXED)")
print("="*60)

# Check TensorFlow version
print(f"TensorFlow Version: {tf.__version__}")

# ---------- 2. SETUP AND CONFIGURATION ----------

IMG_SIZE = 224
BATCH_SIZE = 32
EPOCHS = 10  # Reduced for faster testing
NUM_CLASSES = 2
RANDOM_SEED = 42

np.random.seed(RANDOM_SEED)
tf.random.set_seed(RANDOM_SEED)

print(f"\n Configuration:")
print(f"   Image Size: {IMG_SIZE}x{IMG_SIZE}")
print(f"   Batch Size: {BATCH_SIZE}")
print(f"   Epochs: {EPOCHS}")
print(f"   Classes: {NUM_CLASSES} (Healthy vs Diseased)")

# ---------- 3. DATASET PREPARATION ----------

print("\n" + "="*60)
print("DATASET PREPARATION")
print("="*60)

def create_directory_structure():
    """Create folder structure for dataset"""
    dirs = [
        'dataset/train/healthy',
        'dataset/train/diseased',
        'dataset/test/healthy',
        'dataset/test/diseased',
        'dataset/validation/healthy',
        'dataset/validation/diseased'
    ]
    for dir_path in dirs:
        os.makedirs(dir_path, exist_ok=True)
        print(f" Created: {dir_path}")

create_directory_structure()

print("\n" + "="*50)
print(" DATASET SETUP INSTRUCTIONS")
print("="*50)
print("""
Option 1: Download PlantVillage Dataset from Kaggle
    🔗 https://www.kaggle.com/datasets/abdallahalidev/plantvillage-dataset

Option 2: Use Custom Dataset
    - Organize images as shown below:

    dataset/
    ├── train/
    │   ├── healthy/
    │   │   ├── img1.jpg
    │   │   └── ...
    │   └── diseased/
    │       ├── img1.jpg
    │       └── ...
    ├── test/
    │   ├── healthy/
    │   └── diseased/
    └── validation/
        ├── healthy/
        └── diseased/
""")

# ---------- 4. CREATE SYNTHETIC DATA (No dataset required) ----------

print("\n" + "="*60)
print("GENERATING SYNTHETIC DATA")
print("="*60)

# Generate synthetic images
def generate_synthetic_images(num_images, label):
    """Generate synthetic RGB images for demonstration"""
    images = []
    for _ in range(num_images):
        # Create random image
        img = np.random.randn(64, 64, 3) * 0.5 + 0.5
        # Add pattern based on label
        if label == 0:  # Healthy
            img[20:44, 20:44, :] += 0.3  # Green patch
        else:  # Diseased
            img[10:30, 30:50, :] += 0.4  # Brown patch
            img[40:50, 10:20, :] += 0.3  # Spot
        img = np.clip(img, 0, 1)
        images.append(img)
    return np.array(images)

# Generate training data
print("\n Generating training data...")
train_healthy = generate_synthetic_images(150, 0)
train_diseased = generate_synthetic_images(150, 1)
test_healthy = generate_synthetic_images(30, 0)
test_diseased = generate_synthetic_images(30, 1)

# Combine and label
X_train = np.concatenate([train_healthy, train_diseased])
y_train = np.concatenate([np.zeros(150), np.ones(150)])
X_test = np.concatenate([test_healthy, test_diseased])
y_test = np.concatenate([np.zeros(30), np.ones(30)])

# Resize images to 224x224 for MobileNet
print(" Resizing images...")
X_train_resized = tf.image.resize(X_train, (IMG_SIZE, IMG_SIZE)).numpy()
X_test_resized = tf.image.resize(X_test, (IMG_SIZE, IMG_SIZE)).numpy()

print(f" Generated {len(X_train)} training images")
print(f" Generated {len(X_test)} testing images")

# ---------- 5. DATA AUGMENTATION ----------

print("\n" + "="*60)
print("DATA AUGMENTATION SETUP")
print("="*60)

# Data augmentation for training
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    fill_mode='nearest'
)

test_datagen = ImageDataGenerator(rescale=1./255)

# Create generators
train_generator = train_datagen.flow(
    X_train_resized,
    y_train,
    batch_size=BATCH_SIZE,
    shuffle=True
)

test_generator = test_datagen.flow(
    X_test_resized,
    y_test,
    batch_size=BATCH_SIZE,
    shuffle=False
)

class_names = ['healthy', 'diseased']
num_classes = 2

print(f" Data generators created!")
print(f"   Classes: {class_names}")

# ---------- 6. VISUALIZE SAMPLE IMAGES ----------

print("\n" + "="*60)
print("VISUALIZE SAMPLE IMAGES")
print("="*60)

def plot_sample_images():
    """Plot sample images from the dataset"""
    fig, axes = plt.subplots(2, 5, figsize=(15, 6))

    # Get samples
    healthy_samples = X_train_resized[y_train == 0][:5]
    diseased_samples = X_train_resized[y_train == 1][:5]

    for i in range(5):
        # Healthy
        axes[0, i].imshow(healthy_samples[i])
        axes[0, i].set_title("Healthy")
        axes[0, i].axis('off')

        # Diseased
        axes[1, i].imshow(diseased_samples[i])
        axes[1, i].set_title("Diseased")
        axes[1, i].axis('off')

    plt.suptitle("Sample Training Images", fontsize=14, fontweight='bold')
    plt.tight_layout()
    plt.show()

plot_sample_images()

# ---------- 7. BUILD MODEL ----------

print("\n" + "="*60)
print("BUILDING MOBILENET MODEL")
print("="*60)

def build_model(input_shape, num_classes):
    """Build MobileNet model for transfer learning"""

    # Load pre-trained MobileNet
    base_model = MobileNet(
        weights='imagenet',
        include_top=False,
        input_shape=input_shape
    )

    # Freeze base model layers
    base_model.trainable = False

    # Build complete model
    model = models.Sequential([
        base_model,
        layers.GlobalAveragePooling2D(),
        layers.Dense(128, activation='relu'),
        layers.BatchNormalization(),
        layers.Dropout(0.3),
        layers.Dense(num_classes, activation='softmax')
    ])

    return model

input_shape = (IMG_SIZE, IMG_SIZE, 3)
model = build_model(input_shape, num_classes)

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',  # Changed from categorical_crossentropy
    metrics=['accuracy']
)

print(" Model built successfully!")
model.summary()

# ---------- 8. CALLBACKS ----------

print("\n" + "="*60)
print("SETUP CALLBACKS")
print("="*60)

callbacks = [
    EarlyStopping(
        monitor='val_loss',
        patience=5,
        restore_best_weights=True,
        verbose=1
    ),
    ReduceLROnPlateau(
        monitor='val_loss',
        factor=0.2,
        patience=3,
        min_lr=0.00001,
        verbose=1
    ),
    ModelCheckpoint(
        'best_plant_disease_model.h5',
        monitor='val_accuracy',
        save_best_only=True,
        verbose=1
    )
]

print(" Callbacks configured")

# ---------- 9. TRAIN MODEL ----------

print("\n" + "="*60)
print("TRAINING MODEL")
print("="*60)

# Split training data into train and validation
from sklearn.model_selection import train_test_split

X_train_split, X_val_split, y_train_split, y_val_split = train_test_split(
    X_train_resized, y_train, test_size=0.2, random_state=RANDOM_SEED, stratify=y_train
)

# Create validation generator
val_generator = test_datagen.flow(
    X_val_split,
    y_val_split,
    batch_size=BATCH_SIZE,
    shuffle=False
)

# Reset train generator with training data only
train_generator = train_datagen.flow(
    X_train_split,
    y_train_split,
    batch_size=BATCH_SIZE,
    shuffle=True
)

print(f"\n Starting training...")
print(f"   Training samples: {len(X_train_split)}")
print(f"   Validation samples: {len(X_val_split)}")
print(f"   Epochs: {EPOCHS}")

history = model.fit(
    train_generator,
    steps_per_epoch=len(X_train_split) // BATCH_SIZE,
    epochs=EPOCHS,
    validation_data=val_generator,
    validation_steps=len(X_val_split) // BATCH_SIZE,
    callbacks=callbacks,
    verbose=1
)

print("\n Training completed!")

# Load best model
try:
    model.load_weights('best_plant_disease_model.h5')
    print(" Loaded best model weights")
except:
    print(" Could not load best weights, using current model")

# ---------- 10. EVALUATE MODEL ----------

print("\n" + "="*60)
print("MODEL EVALUATION")
print("="*60)

# Plot training history
def plot_training_history(history):
    """Plot training and validation accuracy/loss"""
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

    # Accuracy
    ax1.plot(history.history['accuracy'], label='Training Accuracy')
    ax1.plot(history.history['val_accuracy'], label='Validation Accuracy')
    ax1.set_title('Model Accuracy')
    ax1.set_xlabel('Epoch')
    ax1.set_ylabel('Accuracy')
    ax1.legend()
    ax1.grid(True)

    # Loss
    ax2.plot(history.history['loss'], label='Training Loss')
    ax2.plot(history.history['val_loss'], label='Validation Loss')
    ax2.set_title('Model Loss')
    ax2.set_xlabel('Epoch')
    ax2.set_ylabel('Loss')
    ax2.legend()
    ax2.grid(True)

    plt.tight_layout()
    plt.show()

plot_training_history(history)

# Evaluate on test data
print("\n Evaluating on test data...")
test_loss, test_accuracy = model.evaluate(test_generator, verbose=0)

# Get predictions
test_generator.reset()
y_pred = model.predict(test_generator)
y_pred_classes = np.argmax(y_pred, axis=1)

print(f"\n Test Accuracy: {test_accuracy:.4f}")
print(f" Test Loss: {test_loss:.4f}")

# Classification Report
print("\n Classification Report:")
print(classification_report(y_test, y_pred_classes, target_names=class_names))

# Confusion Matrix
cm = confusion_matrix(y_test, y_pred_classes)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=class_names, yticklabels=class_names)
plt.title('Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()

# ---------- 11. PREDICTION FUNCTION ----------

print("\n" + "="*60)
print("PREDICTION FUNCTION")
print("="*60)

def predict_plant_disease(image_path, model, class_names, img_size=IMG_SIZE):
    """
    Predict if a plant leaf is healthy or diseased
    """
    try:
        # Load image
        if isinstance(image_path, str):
            img = load_img(image_path, target_size=(img_size, img_size))
        else:
            img = tf.image.resize(image_path, (img_size, img_size)).numpy()
            img = tf.keras.preprocessing.image.array_to_img(img)

        # Preprocess
        img_array = img_to_array(img) / 255.0
        img_array = np.expand_dims(img_array, axis=0)

        # Predict
        predictions = model.predict(img_array, verbose=0)
        predicted_class = np.argmax(predictions[0])
        confidence = np.max(predictions[0]) * 100

        return class_names[predicted_class], confidence

    except Exception as e:
        print(f"Error: {e}")
        return None, 0

# Test prediction
print("\n Testing prediction on sample images:")
for i in range(3):
    test_img = X_test_resized[i]
    pred, conf = predict_plant_disease(test_img, model, class_names)
    actual = class_names[int(y_test[i])]
    print(f"   Sample {i+1}: Actual={actual}, Predicted={pred}, Confidence={conf:.2f}%")

# ---------- 12. BATCH PREDICTION ----------

print("\n" + "="*60)
print("BATCH PREDICTION")
print("="*60)

def batch_predict(folder_path, model, class_names):
    """Predict all images in a folder"""
    results = []

    image_files = [f for f in os.listdir(folder_path)
                   if f.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp', '.tiff'))]

    if not image_files:
        print(f"No images found in '{folder_path}'!")
        return results

    print(f" Processing {len(image_files)} images...")

    for img_file in image_files:
        img_path = os.path.join(folder_path, img_file)
        pred, conf = predict_plant_disease(img_path, model, class_names)

        results.append({
            'image': img_file,
            'prediction': pred,
            'confidence': conf
        })
        print(f"   {img_file}: {pred} ({conf:.2f}%)")

    # Save results
    results_df = pd.DataFrame(results)
    results_df.to_csv('prediction_results.csv', index=False)
    print(f"\n Results saved to 'prediction_results.csv'")

    return results_df

# Create test folder
test_folder = 'test_images'
os.makedirs(test_folder, exist_ok=True)
print(f" Folder '{test_folder}' created for batch prediction")

# ---------- 13. SAVE MODEL ----------

print("\n" + "="*60)
print("SAVING MODEL")
print("="*60)

# Save model
model.save('plant_disease_model.h5')
print(" Model saved as 'plant_disease_model.h5'")

# Save architecture
model_json = model.to_json()
with open('plant_disease_model.json', 'w') as f:
    f.write(model_json)
print(" Architecture saved as 'plant_disease_model.json'")

# TFLite conversion
try:
    converter = tf.lite.TFLiteConverter.from_keras_model(model)
    tflite_model = converter.convert()
    with open('plant_disease_model.tflite', 'wb') as f:
        f.write(tflite_model)
    print(" TFLite model saved as 'plant_disease_model.tflite'")
except Exception as e:
    print(f" TFLite conversion skipped: {e}")

# ---------- 14. SUMMARY ----------

print("\n" + "="*60)
print("FINAL SUMMARY")
print("="*60)

print(f"""
╔══════════════════════════════════════════════════════════════╗
║           PLANT DISEASE DETECTION - TASK 3                   ║
║                   COMPLETED SUCCESSFULLY                     ║
╚══════════════════════════════════════════════════════════════╝

TASKS COMPLETED:

1. DATA GENERATED
   - {len(X_train)} training images
   - {len(X_test)} testing images
   - Classes: {class_names}

2. MODEL BUILT
   - MobileNet (Transfer Learning)
   - Input: {input_shape}

3. MODEL TRAINED
   - Epochs: {EPOCHS}
   - Training: {len(X_train_split)} samples
   - Validation: {len(X_val_split)} samples

4. MODEL EVALUATED
   - Test Accuracy: {test_accuracy:.4f}
   - Test Loss: {test_loss:.4f}

5. PREDICTION SYSTEM
   - Single image prediction
   - Batch prediction
   - Confidence scores

OUTPUT FILES:
   - plant_disease_model.h5
   - plant_disease_model.json
   - plant_disease_model.tflite
   - prediction_results.csv

MODEL PERFORMANCE: {' Good' if test_accuracy > 0.85 else ' Average'}
""")

print("\n COMPLETED SUCCESSFULLY!")
print("="*60)
```
